# Environment Exploration

This notebook explores all 6 environment types available in EnvGen-LLM, demonstrating their observation spaces, action spaces, difficulty levels, and rendering capabilities.

**Environments explored:**
- **PuzzleEnvironment**: Grid-world navigation with items and obstacles
- **MazeEnvironment**: Procedurally generated maze navigation
- **ResourceEnvironment**: Resource collection and allocation
- **TextTaskEnvironment**: Text-based reasoning puzzles
- **CartPoleEnvironment**: Classic CartPole balancing
- **MultiAgentEnvironment**: Two agents on a shared grid

```python
import sys
sys.path.insert(0, '../..')

from envgen.environments.base import EnvironmentConfig
from envgen.utils.types import DifficultyLevel, Action
```

## 1. PuzzleEnvironment

A grid-world where the agent navigates to collect items and reach a goal while avoiding obstacles.

```python
from envgen.environments import PuzzleEnvironment

config = EnvironmentConfig(
    name="exploration_puzzle",
    difficulty=DifficultyLevel.EASY,
    parameters={"grid_size": 8, "num_items": 3, "num_obstacles": 5},
    max_steps=50,
)
env = PuzzleEnvironment(config=config)
env.reset()

print(f"Environment: {config.name}")
print(f"Difficulty: {config.difficulty}")
print(f"Grid size: {config.parameters['grid_size']}")
print(f"Actions: {env.get_available_actions()}")
print(f"Observation keys: {env.get_observation().keys()}")
print()
env.render()
```

## 2. MazeEnvironment

A procedurally generated maze where the agent finds the exit.

```python
from envgen.environments import MazeEnvironment

config = EnvironmentConfig(
    name="exploration_maze",
    difficulty=DifficultyLevel.EASY,
    parameters={"maze_size": 8, "complexity": 0.5},
    max_steps=100,
)
env = MazeEnvironment(config=config)
env.reset()

print(f"Environment: {config.name}")
print(f"Maze size: {config.parameters['maze_size']}")
print(f"Actions: {env.get_available_actions()}")
print(f"Observation keys: {env.get_observation().keys()}")
print()
env.render()
```

## 3. ResourceEnvironment

Resource management where the agent collects and allocates resources under constraints.

```python
from envgen.environments import ResourceEnvironment

config = EnvironmentConfig(
    name="exploration_resource",
    difficulty=DifficultyLevel.EASY,
    parameters={"num_resources": 3, "resource_rate": 1.0},
    max_steps=50,
)
env = ResourceEnvironment(config=config)
env.reset()

print(f"Environment: {config.name}")
print(f"Resource types: {config.parameters['num_resources']}")
print(f"Actions: {env.get_available_actions()}")
print(f"Observation keys: {env.get_observation().keys()}")
print()
env.render()
```

## 4. TextTaskEnvironment

Text-based reasoning puzzles including arithmetic, sequences, logic, and patterns.

```python
from envgen.environments import TextTaskEnvironment

config = EnvironmentConfig(
    name="exploration_text",
    difficulty=DifficultyLevel.EASY,
    parameters={"task_type": "mixed", "num_choices": 4},
    max_steps=10,
)
env = TextTaskEnvironment(config=config)
env.reset()

print(f"Environment: {config.name}")
print(f"Task type: {config.parameters['task_type']}")
print(f"Actions: {env.get_available_actions()}")
print(f"Observation keys: {env.get_observation().keys()}")
print()
env.render()
```

## 5. CartPoleEnvironment

Classic CartPole balancing task with configurable physics.

```python
from envgen.environments import CartPoleEnvironment

config = EnvironmentConfig(
    name="exploration_cartpole",
    difficulty=DifficultyLevel.EASY,
    parameters={"max_steps": 100},
    max_steps=100,
)
env = CartPoleEnvironment(config=config)
env.reset()

print(f"Environment: {config.name}")
print(f"Actions: {env.get_available_actions()}")
print(f"Observation keys: {env.get_observation().keys()}")
print()
env.render()
```

## 6. MultiAgentEnvironment

Two agents on a shared grid competing or cooperating to collect items.

```python
from envgen.environments import MultiAgentEnvironment

config = EnvironmentConfig(
    name="exploration_multi_agent",
    difficulty=DifficultyLevel.EASY,
    parameters={"grid_size": 8, "num_items": 5, "cooperation_mode": False},
    max_steps=50,
)
env = MultiAgentEnvironment(config=config)
env.reset()

print(f"Environment: {config.name}")
print(f"Grid size: {config.parameters['grid_size']}")
print(f"Cooperation mode: {config.parameters['cooperation_mode']}")
print(f"Actions: {env.get_available_actions()}")
print(f"Observation keys: {env.get_observation().keys()}")
print()
env.render()
```

## Summary

All 6 environments support:
- `reset()` — Reset the environment to initial state
- `step(action)` — Take an action and return (next_obs, reward, done, info)
- `get_observation()` — Get current observation as a dict
- `get_available_actions()` — Get list of available actions
- `render()` — Visualize the current state
- Configurable difficulty levels: EASY, MEDIUM, HARD, EXPERT
